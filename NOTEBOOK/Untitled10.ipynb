```python
import pandas as pd
file_path = '/content/drive/MyDrive/corporate_ai_adoption_dataset.csv'
df = pd.read_csv(file_path)
print(df.describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

plt.figure(figsize=(12, 5))

# 1. Left Plot: Distribution
plt.subplot(1, 2, 1)
sns.histplot(df['ai_adoption_level'], kde=True, color='purple')
plt.title('AI Adoption Level Distribution')

# 2. Right Plot: Heatmap
plt.subplot(1, 2, 2)
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')

plt.tight_layout()
plt.savefig('relationship_plots.png')
plt.show()
```

```python
high_maturity = df[df['ai_maturity_score'] >= 5.0]['ai_adoption_level']
low_maturity = df[df['ai_maturity_score'] < 5.0]['ai_adoption_level']

t_stat, p_val = stats.ttest_ind(high_maturity, low_maturity, equal_var=False)

print("\n--- Phase 5: Hypothesis Testing ---")
print("T-Statistic:", t_stat)
print("P-Value:", p_val)
```
